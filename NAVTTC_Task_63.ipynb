## Step 1 — CNN model on small image dataset

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train = x_train[:10000].astype('float32') / 255.0
x_test = x_test[:2000].astype('float32') / 255.0
y_train = y_train[:10000]
y_test = y_test[:2000]
x_train = x_train[..., None]
x_test = x_test[..., None]
model = models.Sequential([layers.Input(shape=(28, 28, 1)), layers.Conv2D(16, 3, activation='relu'), layers.MaxPooling2D(), layers.Flatten(), layers.Dense(10, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=2, validation_split=0.1, verbose=1)
print('Test evaluation:', model.evaluate(x_test, y_test, verbose=0))
```
