## Step 1 — Copy and view

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import numpy as np
original = np.array([1, 2, 3])
copy_arr = original.copy()
view_arr = original.view()
original[0] = 99
print('original:', original)
print('copy:', copy_arr)
print('view:', view_arr)
```
