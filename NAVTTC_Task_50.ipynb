## Step 1 — Standard deviation

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import statistics
numbers = [2, 4, 4, 4, 5, 5, 7, 9]
print('Sample standard deviation:', round(statistics.stdev(numbers), 3))
print('Population standard deviation:', round(statistics.pstdev(numbers), 3))
```
