## Step 1 — Handle missing values

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import pandas as pd
df = pd.DataFrame({'name': ['Ali', 'Sara', 'Ahmed'], 'marks': [80, None, 75]})
print('Missing values before:')
print(df.isna().sum())
df['marks'] = df['marks'].fillna(df['marks'].mean())
print('After filling:')
print(df)
```
