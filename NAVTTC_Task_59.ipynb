## Step 1 — Mean squared error

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import numpy as np
y_true = np.array([3.0, -0.5, 2.0, 7.0])
y_pred = np.array([2.5, 0.0, 2.1, 7.8])
mse = np.mean((y_true - y_pred) ** 2)
print('MSE:', round(mse, 3))
print('RMSE:', round(np.sqrt(mse), 3))
```
