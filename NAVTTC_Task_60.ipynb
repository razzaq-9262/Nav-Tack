## Step 1 — Text classification example

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
texts = ['win cash prize now', 'claim your free reward', 'meeting is at noon', 'please send the report']
labels = ['spam', 'spam', 'normal', 'normal']
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)
model = MultinomialNB().fit(X, labels)
new_text = ['win a free prize']
print('Prediction:', model.predict(vectorizer.transform(new_text))[0])
```
