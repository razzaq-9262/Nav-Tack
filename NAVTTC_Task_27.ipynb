## Step 1 — Custom exception

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
class NegativeAgeError(Exception):
    pass

age = -2
try:
    if age < 0:
        raise NegativeAgeError('Age cannot be negative')
    print(age)
except NegativeAgeError as error:
    print(error)
```
