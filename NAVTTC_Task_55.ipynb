## Step 1 — Create a time series

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import pandas as pd
import matplotlib.pyplot as plt
dates = pd.date_range('2025-01-01', periods=12, freq='MS')
values = [10, 12, 11, 14, 16, 15, 18, 20, 19, 22, 24, 25]
series = pd.Series(values, index=dates)
print(series)
series.plot(title='Monthly Values')
plt.xlabel('Month')
plt.ylabel('Value')
plt.show()
```
