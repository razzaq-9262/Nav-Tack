## Step 1 — Normalize text

Beginner tip: run this cell and look at each line. Variables are labelled boxes; functions perform a job; `print()` shows the result.

```python
import re
text = 'Hello!!! This is a SAMPLE message, with numbers 123.'
clean = re.sub(r'[^a-zA-Z\s]', '', text.lower())
print(clean)
print('Tokens:', clean.split())
```
