## Step 3 — Check common AI libraries

```python
for package in ['numpy', 'pandas', 'sklearn', 'torch', 'tensorflow']:
    try:
        module = __import__(package)
        print(package, getattr(module, '__version__', 'installed'))
    except Exception as e:
        print(package, 'not available:', type(e).__name__)
```
